```python
import json

text_file_path = ["/content/drive/MyDrive/medmoe/blood_heart_circulation_qa.json",
             "/content/drive/MyDrive/medmoe/bones_joints_muscles_qa.json",
             "/content/drive/MyDrive/medmoe/mental_behavior_qa.json"]

text_data = []

for f in text_file_path:
    with open(f, "r") as file:
        qa_data = json.load(file)
        for topic in qa_data:
            for question, answer in topic['question_answer_pair']:
                text_data.append(question)
                text_data.append(answer)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("TsinghuaC3I/Llama-3-8B-UltraMedical")
```

```python
from collections import Counter
import re

def clean_token(token):
    return re.sub(r'[^\w\-]', '', token.lower())  # keep hyphens like "T-cell"

word_counts = Counter()
split_counts = Counter()

for text in text_data:
    for word in text.split():
        word = clean_token(word)
        if not word or word.isnumeric():
            continue
        word_counts[word] += 1
        tokens = tokenizer.tokenize(word)
        if len(tokens) > 1:
            split_counts[word] += 1

candidates = [word for word, count in split_counts.items() if len(word) >= 1 and count >= 1]
```

```python
tokenizer.add_tokens(candidates)
```

```python
tokenizer.save_pretrained("/content/drive/MyDrive/medmoe/shared_tokenizer/")
```
